# As(III) m6A inhibition — analysis notebook

Loads `results/composite_ranking.tsv` and accompanying tables to:
1. Plot family-level distribution of composite scores.
2. Sanity-check controls (PIN1, TXN1, GAPDH should rank high; SH3 should rank low).
3. Render per-target Cys-landscape views for the top hits.
4. Tabulate inference tiers per family (likely_inhibitory / possibly_allosteric / binding_only / no_binding).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

RESULTS = Path('../results')
rank = pd.read_csv(RESULTS / 'composite_ranking.tsv', sep='\t')
cys = pd.read_csv(RESULTS / 'cys_table.tsv', sep='\t')
cov = pd.read_csv(RESULTS / 'docking_covalent.tsv', sep='\t')
rank.head(10)
```

## Family-level distribution

```python
fig, ax = plt.subplots(figsize=(8, 4))
sns.stripplot(data=rank, x='family', y='composite_score', hue='inference_tier', size=10, ax=ax)
for _, r in rank.iterrows():
    ax.text(list(rank['family'].unique()).index(r['family']) + 0.1, r['composite_score'], r['gene'], fontsize=8)
ax.set_title('Composite inhibition-likelihood score by family')
plt.tight_layout()
```

## Control sanity check

```python
controls = rank[rank['family'].isin(['control_pos', 'control_neg'])]
controls.sort_values('composite_score', ascending=False)
```

## Inference tier breakdown

```python
rank.groupby(['family', 'inference_tier']).size().unstack(fill_value=0)
```
